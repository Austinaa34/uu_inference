#### Question 7:

```python
import numpy as np
import pandas as pd

df = pd.read_csv("../data/ames_prices.csv")
df["central_air"] = (df["Central.Air"] == "Y").astype(int)
x = df["central_air"]
n = len(x)

# a: 
p_hat = x.mean()
print(p_hat)
```

```python
# b:

seed, B = 42, 2000
rng = np.random.default_rng(seed)
boot = np.array([rng.choice(x, size=n, replace=True).mean() for _ in range(B)])
```

```python
# c:
se = boot.std(ddof=1)
print(se)
```

```python
# d:
lo_90 = p_hat - 1.645 * se
hi_90 = p_hat + 1.645 * se

lo_95 = p_hat - 1.96 * se
hi_95 = p_hat + 1.96 * se

lo_99 = p_hat - 2.576 * se
hi_99 = p_hat + 2.576 * se

print(f"90% CI: ({lo_90:.3f}, {hi_90:.3f})")
print(f"95% CI: ({lo_95:.3f}, {hi_95:.3f})")
print(f"99% CI: ({lo_99:.3f}, {hi_99:.3f})")
```

```python
# e:

p0 = 0.90

print("alpha = 0.10:", "fail to reject the null hypothesis" if lo_90 <= p0 <= hi_90 
      else "reject the null hypothesis")
print("alpha = 0.05:", "fail to reject the null hypothesis" if lo_95 <= p0 <= hi_95 
      else "reject the null hypothesis")
print("alpha = 0.01:", "fail to reject the null hypothesis" if lo_99 <= p0 <= hi_99 
      else "reject the null hypothesis")
```

**None of the confidence intervals contain 0.90 so they all reject the null hypothesis.**

```python
x_q8 = np.arcsinh(df["area"])
y_q8 = np.arcsinh(df["price"])
n_q8 = len(x_q8)

orig_est = np.corrcoef(x_q8, y_q8)[0, 1]
print(orig_est)
```

```python
# a:
# 
# Used ai to help with the for loop. 
seed, B = 42, 2000
rng = np.random.default_rng(seed)
boot = np.empty(B)
for b in range(B):
    idx = rng.integers(0, n_q8, size=n_q8)
    boot[b] = np.corrcoef(x_q8[idx], y_q8[idx])[0, 1]
```

```python
# b:

se = boot.std(ddof=1)
print(se)
```

```python
# c:

lo_90, hi_90 = orig_est - 1.645 * se, orig_est + 1.645 * se
lo_95, hi_95 = orig_est - 1.96 * se,  orig_est + 1.96 * se
lo_99, hi_99 = orig_est - 2.576 * se, orig_est + 2.576 * se

print(f"90% CI: ({lo_90:.3f}, {hi_90:.3f})")
print(f"95% CI: ({lo_95:.3f}, {hi_95:.3f})")
print(f"99% CI: ({lo_99:.3f}, {hi_99:.3f})")
```

```python
# d: 

print("alpha = 0.10:", "fail to reject the null hypothesis" if lo_90 <= p0 <= hi_90 
      else "reject the null hypothesis")
print("alpha = 0.05:", "fail to reject the null hypothesis" if lo_95 <= p0 <= hi_95 
      else "reject the null hypothesis")
print("alpha = 0.01:", "fail to reject the null hypothesis" if lo_99 <= p0 <= hi_99 
      else "reject the null hypothesis")
```

**Since the null hypothesis is rejected, and the CI are betwen 0.697-0.75, this means that as area increases, price increases.**

```python
metabric = pd.read_csv("../data/metabric.csv")

x_q9 = metabric["Overall Survival (Months)"]
n_q9 = len(x_q9)

mean_hat = x_q9.mean()
print(mean_hat)
```

```python
# a:

seed, B = 42, 2000
rng = np.random.default_rng(seed)
boot = np.empty(B)
for b in range(B):
    boot[b] = rng.choice(x_q9, size=n_q9, replace=True).mean()
```

```python
# b: 

se = boot.std(ddof=1)
print(se)
```

```python
# c:

lo_90, hi_90 = mean_hat - 1.645 * se, mean_hat + 1.645 * se
lo_95, hi_95 = mean_hat - 1.96 * se,  mean_hat + 1.96 * se
lo_99, hi_99 = mean_hat - 2.576 * se, mean_hat + 2.576 * se

print(f"90% CI: ({lo_90:.3f}, {hi_90:.3f})")
print(f"95% CI: ({lo_95:.3f}, {hi_95:.3f})")
print(f"99% CI: ({lo_99:.3f}, {hi_99:.3f})")
```

```python
# d:
u = 120

print("alpha = 0.10:", "fail to reject the null hypothesis" if lo_90 <= u <= hi_90 
      else "reject the null hypothesis")
print("alpha = 0.05:", "fail to reject the null hypothesis" if lo_95 <= u <= hi_95 
      else "reject the null hypothesis")
print("alpha = 0.01:", "fail to reject the null hypothesis" if lo_99 <= u <= hi_99 
      else "reject the null hypothesis")
```

**e:** The estimate is 129 months (10.8 years) and the confidence intervals have it between 123 to 134 months (10.25 - 11.17 years). The results do not estimate the unrestricted mean lifetime because the living patients are only recording their last follow up.

```python
# a:

seed, B = 42, 2000
rng = np.random.default_rng(seed)
boot = np.empty(B)
for b in range(B):
    boot[b] = rng.choice(x10, size=n, replace=True).mean()
```

```python
# b:

se = boot.std(ddof=1)
print(se)
```

```python
# c:

lo_90, hi_90 = mean_hat - 1.645 * se, mean_hat + 1.645 * se
lo_95, hi_95 = mean_hat - 1.96 * se,  mean_hat + 1.96 * se
lo_99, hi_99 = mean_hat - 2.576 * se, mean_hat + 2.576 * se

print(f"90% CI: ({lo_90:.3f}, {hi_90:.3f})")
print(f"95% CI: ({lo_95:.3f}, {hi_95:.3f})")
print(f"99% CI: ({lo_99:.3f}, {hi_99:.3f})")
```

```python
# d:

mu = 8

print("alpha = 0.10:", "fail to reject the null hypothesis" if lo_90 <= mu <= hi_90 
      else "reject the null hypothesis")
print("alpha = 0.05:", "fail to reject the null hypothesis" if lo_95 <= mu <= hi_95 
      else "reject the null hypothesis")
print("alpha = 0.01:", "fail to reject the null hypothesis" if lo_99 <= mu <= hi_99 
      else "reject the null hypothesis")
```

**e:** The statistical evidence shows that the respondents were getting less than 8 hours of sleep per night. This is because the initial mean estimate is 7.62 and the confidence intervals are below 8 as well. From a practical standpoint, the difference is around 23 minutes which is not that much a difference. For example, if I were to lose 23 minutes of sleep, it would not have much of an effect on me compared to if I were to lose an hour of sleep. The practical siginficance differs from the statistical signficance which is indicating that it is a meaningful difference.
